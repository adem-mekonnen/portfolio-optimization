```python
import yfinance as yf
import pandas as pd
import os
```

```python
# Step 1: Define tickers and dates
tickers = ['TSLA', 'BND', 'SPY']
start_date = '2015-01-01'
end_date = '2026-01-15'
```

```python
# Step 2: Ensure the directory exists (Prevent FileNotFoundError)
os.makedirs('data/raw', exist_ok=True)
```

```python
# Step 3: Download data with auto_adjust=False
# This tells yfinance: "Don't hide the Adj Close column, give it to me explicitly."
raw_data = yf.download(tickers, start=start_date, end=end_date, auto_adjust=False)
```

```python
# Step 4: Select 'Adj Close'
# Because we used auto_adjust=False, this column now exists.
data = raw_data['Adj Close']
```

```python
# Step 5: Save to CSV
data.to_csv('data/raw/historical_data.csv')

print("Success! Data saved to data/raw/historical_data.csv")
print(data.head())
```
